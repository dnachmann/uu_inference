## Problem 1

```python
xbar = 52
s = 10
n = 100

xbar_se = s / np.sqrt(n)
print(xbar_se)


z = (xbar - 50) / xbar_se
print(z)

lower_bound, upper_bound = stats.norm.interval(
    confidence = .95,
    loc = xbar,
    scale = xbar_se
) 

print(lower_bound, upper_bound)
```

We reject the null hypothesis that mu = 50 at alpha = 0.05, and we fail to reject the null hypothesis at alpha = 0.01. The 95% confidence interval for the true population mean is [50.040, 53.959].

## Problem 7

```python
df = pd.read_csv('data/ames_prices.csv',index_col = 0)
df = pd.get_dummies(df, columns = ['Central.Air'], dtype = int, drop_first=True)
df['Central.Air_Y']
```

```python
# a
print(df.value_counts(df['Central.Air_Y']))
print(f'Proportion of homes with central air: {2734 / len(df)}')
p = 2734 / len(df)
```

```python
# c
random.seed(50)
rng = np.random.default_rng(seed=50)

n = 5000
props = []

for i in range(n):
    sample = rng.choice(df['Central.Air_Y'], size=len(df), replace=True)
    prop_i = np.mean(sample)
    props.append(prop_i)

print(f'The seed is 50 and the number of bootstrap replications is {n}.')
```

```python
# c
se = np.std(props, ddof = 1)
print(se)
```

```python
# d

lower_bound_90, upper_bound_90 = stats.norm.interval(
    confidence = .90,
    loc = p,
    scale = se 
) 

print(f'90% confidence interval: {float(lower_bound_90):.6f}, {float(upper_bound_90):.6f}')

lower_bound_95, upper_bound_95 = stats.norm.interval(
    confidence = .95,
    loc = p,
    scale = se 
) 

print(f'95% confidence interval: {float(lower_bound_95):.6f}, {float(upper_bound_95):.6f}')

lower_bound_99, upper_bound_99 = stats.norm.interval(
    confidence = .99,
    loc = p,
    scale = se 
) 

print(f'99% confidence interval: {float(lower_bound_99):.6f}, {float(upper_bound_99):.6f}')
```

f. 
H0: p = 0.9, H1: p != 0.9

None of the bootstrap confidence intervals contain 0.9, so we reject the null hypothesis at significance levels alpha = 0.10, 0.05, 0.01. Given these results, we have sufficient evidence to conclude that the true population proportion of houses with central air is significantly different from 90%.

The 99% boostrap confidence interval is [0.921018, 0.945193]. Based on this, we are 99% confident the true population proportion of houses with central air lies between 92.1018% and 94.5193%.

## Problem 8

```python
area = np.arcsinh(df['area'])
price = np.arcsinh(df['price'])

correlation_coefficient = np.corrcoef(area, price)[0,1]
print(correlation_coefficient)
```

```python
# a 
random.seed(50)
rng = np.random.default_rng(seed=50)

n = 5000
corrs = []

for i in range(n):
    index = rng.choice(len(df), size=len(df), replace=True)
    area = np.arcsinh(df['area'].to_numpy()[index])
    price = np.arcsinh(df['price'].to_numpy()[index])
    corr_i = np.corrcoef(area, price)[0, 1]
    corrs.append(corr_i)

print(f'The seed is 50 and the number of bootstrap replications is {n}.')
```

```python
# b
se = np.std(corrs, ddof = 1)
print(se)
```

```python
# c
lower_bound_90, upper_bound_90 = stats.norm.interval(
    confidence = .90,
    loc = correlation_coefficient,
    scale = se 
) 

print(f'90% confidence interval: {float(lower_bound_90):.6f}, {float(upper_bound_90):.6f}')

lower_bound_95, upper_bound_95 = stats.norm.interval(
    confidence = .95,
    loc = correlation_coefficient,
    scale = se 
) 

print(f'95% confidence interval: {float(lower_bound_95):.6f}, {float(upper_bound_95):.6f}')

lower_bound_99, upper_bound_99 = stats.norm.interval(
    confidence = .99,
    loc = correlation_coefficient,
    scale = se 
) 

print(f'99% confidence interval: {float(lower_bound_99):.6f}, {float(upper_bound_99):.6f}')
```

d. 

Because 0 is not in the 90% confidence interval, at a 90% confidence level we have statistically significant evidence to conclude that rho != 0 and we reject the null hypothesis.

Because 0 is not in the 95% confidence interval, at a 95% confidence level we have statistically significant evidence to conclude that rho != 0 and we reject the null hypothesis.

Because 0 is not in the 99% confidence interval, at a 99% confidence level we have statistically significant evidence to conclude that rho != 0 and we reject the null hypothesis.

e. 

Area and price must remain paired during resampling because they are highly correlated with a correlation coefficient of 72.3%. If the variables were not paired, the correlation between them would not be reflected in the bootstrapped samples.

```python
df = pd.read_csv('data/metabric.csv')
avg_months = df['Overall Survival (Months)'].mean()
print(f'{avg_months:.3f}')
```

```python
# a

random.seed(50)
rng = np.random.default_rng(seed=50)

n = 5000
means = []

for i in range(n):
    sample = rng.choice(df['Overall Survival (Months)'], size=len(df), replace=True)
    mean_i = np.mean(sample)
    means.append(mean_i)

print(f'The seed is 50 and the number of bootstrap replications is {n}.')
```

```python
# b
se = np.std(means, ddof = 1)
print(se)
```

```python
#c 

lower_bound_90, upper_bound_90 = stats.norm.interval(
    confidence = .90,
    loc = avg_months,
    scale = se 
) 

print(f'90% confidence interval: {float(lower_bound_90):.3f}, {float(upper_bound_90):.3f}')

lower_bound_95, upper_bound_95 = stats.norm.interval(
    confidence = .95,
    loc = avg_months,
    scale = se 
) 

print(f'95% confidence interval: {float(lower_bound_95):.3f}, {float(upper_bound_95):.3f}')

lower_bound_99, upper_bound_99 = stats.norm.interval(
    confidence = .99,
    loc = avg_months,
    scale = se 
) 

print(f'99% confidence interval: {float(lower_bound_99):.3f}, {float(upper_bound_99):.3f}')
```

d.

Because 120 is not in any of the confidence intervals, at all three confidence levels 90%, 95% and 99%, we have significant evidence to conclude that the true population mean is not equal to 120, and we reject the null hypothesis.

e. 

The estimate 129.094 means that on average, patients live for an average of 129.094 months, or 10.758 years after surgery. 

The 90% confidence interval states that we are 90% confident the true population average survivial time is between 125.586 months (10.466 years) and 132.603 months (11.050 years). 

The 95% confidence interval estimates the true population mean survival time to be between 124.913 months (10.409 years) and 133.275 months (11.106 years).

The 99% confidence interval estimates the average population survival time after surgery between 123.600 months (10.300 years) and 134.589 months (11.216 years).

Because the results are right-censored, meaning some patients were still alive when the last observation of the study concluded, their true length of survival is unknown. This right-censorship means the results do not estimate unrestricted mean lifetime, which would only be estimated if all patients in the estimate were dead at the end of the study.

```python
df = pd.read_csv('data/nhanes_data_17_18.csv', low_memory=False)
df = df[df['AgeInYearsAtScreening'] > 18]
sleep_mean = df['SleepHoursWeekdaysOrWorkdays'].dropna().mean()
print(sleep_mean)
```

```python
# a
random.seed(50)
rng = np.random.default_rng(seed=50)

n = 5000
means = []

for i in range(n):
    sample = rng.choice(df['SleepHoursWeekdaysOrWorkdays'].dropna(), size=len(df), replace=True)
    mean_i = np.mean(sample)
    means.append(mean_i)

print(f'The seed is 50 and the number of bootstrap replications is {n}.')
```

```python
# b
se = np.std(means, ddof = 1)
print(se)
```

```python
lower_bound_90, upper_bound_90 = stats.norm.interval(
    confidence = .90,
    loc = sleep_mean,
    scale = se 
) 

print(f'90% confidence interval: {float(lower_bound_90):.3f}, {float(upper_bound_90):.3f}')

lower_bound_95, upper_bound_95 = stats.norm.interval(
    confidence = .95,
    loc = sleep_mean,
    scale = se 
) 

print(f'95% confidence interval: {float(lower_bound_95):.3f}, {float(upper_bound_95):.3f}')

lower_bound_99, upper_bound_99 = stats.norm.interval(
    confidence = .99,
    loc = sleep_mean,
    scale = se 
) 

print(f'99% confidence interval: {float(lower_bound_99):.3f}, {float(upper_bound_99):.3f}')
```

d. 

At a significance level of 0.1, we reject the null hypothesis that the population average hours of sleep is equal to 8.

At significance level 0.05, we also reject the null hypothesis that the population average hours of sleep is equal to 8.

With a significance level of 0.01 we reject the null hypothesis that the population average hours of sleep is equal to 8.

e. 

Despite the results being statistically significant, the practical significance can be called into question. An average sleep time of 7.547 hours, the lower bound for the 99% confidence interval, is only about 27 fewer minutes of sleep than the null hypothesis of 8 hours, which is presumably not enough to make a major difference.
